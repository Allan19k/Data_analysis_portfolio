```python
# Librerias necesarias
import requests # hacer la solicitud web
from bs4 import BeautifulSoup # parsear el html
import matplotlib.pyplot as plt # graficar los resultados
```

```python
# url: contiene la página de Hacker News con un hilo de discusión.
url = "https://news.ycombinator.com/item?id=29782099"
response = requests.get(url) # descarga el contenido de la página.
```

```python
# Se procesa la página web
soup = BeautifulSoup(response.content, "html.parser") # crea un objeto para recorrer el HTML.
elements = soup.find_all(class_="ind", indent=0) # busca elementos con la clase "ind" (estructura de los comentarios en Hacker News).
comments = [e.find_next(class_="comment") for e in elements] # localiza los textos de cada comentario
```

```python
# Se define un diccionario con lenguajes de programación como claves, iniciando cada uno en 0 menciones.
keywords = {
        "python": 0,
        "javascript": 0,
        "typescript": 0,
        "ruby": 0,
        "java": 0,
        "rust": 0,
        "c#": 0,
    }
```

```python
# se analiza cada comentario
for comment in comments:
        comment_text = comment.get_text().lower() # convierte todo el texto a minúsculas. Para no tener problemas en caso de que la palabra inicie con mayúscula o sean puras mayúsculas
        words = comment_text.split(" ") # divide en palabras
        words = {w.strip(".,/:;!@") for w in words} # elimina puntuacion innecesaria

        for k in keywords:
            if k in words:
                keywords[k] += 1 #Se revisa si alguna palabra coincide con los lenguajes en keywords. Si hay coincidencia, se incrementa el contador del lenguaje.

print(keywords)
```

```python
plt.figure(figsize=(12, 6)) # Increase figure size
plt.bar(keywords.keys(), keywords.values()) # Se grafica un diagrama de barras mostrando cuántas veces se mencionó cada lenguaje en los comentarios.
plt.xlabel("Language")
plt.ylabel("# of Mentions")
plt.xticks(rotation=45, ha="right") # Rotate labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```
