```python
users = [
    {"id": 0, "name": "Hero"},
    {"id": 1, "name": "Dunn}"},
    {"id": 2, "name": "Sue"},
    {"id": 3, "name": "Chi"},
    {"id": 4, "name": "Thor"},
    {"id": 5, "name": "Clive"},
    {"id": 6, "name": "Hicks"},
    {"id": 7, "name": "Devin"},
    {"id": 8, "name": "Kate"},
    {"id": 9, "name": "Klein"}
]
```

```python
friendship_pairs = [(0,1),(0,2),(1,2),(1,3),(2,3),(3,4),
                    (4,5),(5,6),(5,7),(6,8),(7,8),(8,9)]
```

```python
friendships = {user["id"]: [] for user in users}
#loop
for i,j in friendship_pairs:
  friendships[i].append(j) # add j as a friend of user i
  friendships[j].append(i) #add i as a friend of user j
```

```python
def number_of_friends(user):
  user_id = user["id"]
  friend_ids = friendships[user_id]
  return len(friend_ids)
total_connections = sum(number_of_friends(user) for user in users)
```

```python
num_users = len(users) #lenght of users list
avg_connections = total_connections / num_users
```

```python
# create a list (user_id, number_of_friends)
num_friends_by_id = [(user["id"], number_of_friends(user)) for user in users]
```

```python
num_friends_by_id.sort( #sort the list by num of friends largest to smallest
    key= lambda id_and_friends: id_and_friends[1],
    reverse=True)
```

```python
def foaf_ids_bad(user):
  #foaf is short for "friend of a friend"
  return [foaf_id
          for friend_id in friendships[user["id"]]
          for foaf_id in friendships[friend_id]]
```

```python
from collections import Counter
def friends_of_friends(user):
  user_id = user["id"]
  return Counter(
      foaf_id
      for friend_id in friendships[user_id]
      for foaf_id in friendships[friend_id]
      if foaf_id != user_id
      and foaf_id not in friendships[user_id]
  )

print(friends_of_friends(users[3]))
```

#5. ¿Cuales son los 3 empleados que tiene mejor salario?, ¿Que tienen en común?

## Combinar datos

Combinar los datos de  `users` con `salaries_and_tenures` para que cada usuario tenga el salario y su tenencia


```python
user_salary_tenure = []
for i, user in enumerate(users):
  salary, tenure = salaries_and_tenures[i]
  user_salary_tenure.append({
      "id": user["id"],
      "nombre": user["name"],
      "salario": salary,
      "tenura": tenure
  })

display(user_salary_tenure)
```

## Organizar por salario

Ordenar los datos por salario en orden descendente para obtener los empleados con el salario más alto


```python
user_salary_tenure.sort(key=lambda x: x['salario'], reverse=True)
display(user_salary_tenure)
```

```python
top_3_employees = user_salary_tenure[:3]
display(top_3_employees)
```

## Obtener los intereses

Obtener los intereses de los 3 empleados con el salario mas alto de  `interests`


```python
top_3_employee_interests = []
for employee in top_3_employees:
  employee_id = employee["id"]
  for interest in interests:
    if interest[0] == employee_id:
      top_3_employee_interests.append(interest)

display(top_3_employee_interests)
```

## Obtener intereses comúnes


Analizar los intereses de cada uno de los 3 empleados y ver que tienen en común  


```python
interest_strings = [interest[1] for interest in top_3_employee_interests]
interest_counts = Counter(interest_strings)

common_interests = {interest: count for interest, count in interest_counts.items() if count > 1}

print("Intereses comunes entre los 3 empleados:")
for interest, count in common_interests.items():
    print(f"{interest}: {count}")
```

```python

print("Los 3 empleados con el salario más alto:")

for employee in top_3_employees:
    print(f"Nombre: {employee['nombre']}, ${employee['salario']}")

print("\nIntreses en común:")

if common_interests:
    for interest, count in common_interests.items():
        print(f"{interest}: {count}")
else:
    print("No se encontraron intereses en común.")
```
